In this exercise student familiarizes the neural network classification problem.
The purpose is use MLP classification in Penguin  Dataset to classify different species just like in the Iris dataset. The dataset contains measurement of 3 different species of penguins coming from 3 islands in the Palmer Archipelago. These three classes are Adelie, Chinstrap, and Gentoo. If ‘Gentoo’ sounds familiar that is because Gentoo Linux is named after it! Also, these datasets contain culmen dimensions for each species. The culmen is the upper ridge of a bird’s bill. 

Columns in dataset:
- __species__ a factor denoting penguin species (Adélie, Chinstrap and Gentoo)
- __island__ a factor denoting island in Palmer Archipelago, Antarctica (Biscoe, Dream or Torgersen)
- __bill_length_mm__ a number denoting bill length (millimeters)
- __bill_depth_mm__ a number denoting bill depth (millimeters)
- __flipper_length_mm__ an integer denoting flipper length (millimeters)
- __body_mass_g__ an integer denoting body mass (grams)
- __sex__ a factor denoting penguin sex (female, male)

```python
# Import libraries and dataset
# Implementation comes this cell

# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

import tensorflow as tf
from tensorflow.keras import Input, Model
from tensorflow.keras.layers import Dense

# Load the dataset and display the first few rows
df_penguins = pd.read_csv('penguins_size.csv')
print(df_penguins.head())
```

```python
# Data examination (You can do as you want)
# Implementation comes this cell

# Get basic information about the dataset
df_penguins.info() 

# Check for unique values in categorical columns
for col in ['species', 'island', 'sex']:
    print(col, df_penguins[col].unique())

# Treat the invalid '.' in sex as missing, then drop rows with any null
df_penguins['sex'] = df_penguins['sex'].replace('.', np.nan)

# Check for missing values and drop rows with any null values
print(df_penguins.isnull().sum())
df_penguins = df_penguins.dropna().reset_index(drop=True)
print(df_penguins.isnull().sum())

# Get the shape of the dataset after cleaning
print("Shape after cleaning:", df_penguins.shape)
```

```python
# Categorical varibles: Encode columns Species, Island and Sex
# species = Adélie, Chinstrap and Gentoo
# sex = MALE, FEMALA
# island = Torgersen, Biscoe and Dream

# Features = island, culmen_length_mm,culmen_depth_mm, flipper_length_mm, body_mass_g & sex
# Label = species

# Implementation comes this cell

# Group numeric and categorical features separately for preprocessing
categorical_cols    = ["island", "sex"]
numeric_cols    = ["culmen_length_mm", "culmen_depth_mm", "flipper_length_mm", "body_mass_g"]

# Convert categorical features to dummy variables
dummies_categorical_df = pd.get_dummies(df_penguins[categorical_cols], dtype=int)

# Scale all the numeric features using StandardScaler
scaler_x = StandardScaler()
scaled_numeric = scaler_x.fit_transform(df_penguins[numeric_cols]) # returns NumPy array

# Prepare features (X)
X_scaled = np.concatenate((scaled_numeric, dummies_categorical_df.to_numpy()), axis=1)

# Prepare target variable (y)
enc = LabelEncoder()
y_labeled = enc.fit_transform(df_penguins["species"])
```

```python
# Dataset splitting
# training set 80%
# test set 20%
# Implementation comes this cell

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y_labeled, test_size=0.2, random_state=42, stratify=y_labeled
)
```

```python
# Evaluate()
# Implementation comes this cell

n_features = X_train.shape[1]   # 9
n_classes = len(enc.classes_)   # 3

layer_nodes_combo_options = [[2], [4], [8], [16], [4, 2], [8, 4]]
epoch_options = [10, 50, 100, 200]
n_runs = 5   # train each config 5 times with different seeds

results_table = []

for layer_nodes_combo in layer_nodes_combo_options:
    for n_epochs in epoch_options:
        test_acc_runs = []   # this config's test accuracy across all runs

        for run in range(n_runs):
            tf.keras.utils.set_random_seed(run)   # seeds 0, 1, 2, 3, 4

            inputs = Input(shape=(n_features,))
            x = inputs
            for n_nodes in layer_nodes_combo:
                x = Dense(n_nodes, activation='relu')(x)
            outputs = Dense(n_classes, activation='softmax')(x)
            model = Model(inputs=inputs, outputs=outputs)
            model.compile(optimizer='adam',
                          loss='sparse_categorical_crossentropy',
                          metrics=['accuracy'])

            model.fit(X_train, y_train, epochs=n_epochs, verbose=0)
            test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
            test_acc_runs.append(test_acc)

        avg_acc = np.mean(test_acc_runs)
        std_acc = np.std(test_acc_runs)

        results_table.append((layer_nodes_combo, n_epochs, avg_acc, std_acc))
        print(f"Layer/Nodes: {layer_nodes_combo}, Epochs: {n_epochs} → "
              f"Avg Test Accuracy: {avg_acc:.4f} (± {std_acc:.4f} across {n_runs} runs)")
```

## Write conclusion:
### How number of nodes in hidden layer affects the results?
More nodes made the network learn faster. With only 10 epochs, the network with 2 nodes got about 44% accuracy, but the network with 16 nodes got about 92%. The 16-node network reached 100% accuracy after only 50 epochs. 
A small network can still learn well, but it needs more training. The 2-node network also reached 100% accuracy, but only after 200 epochs. 
I also tried networks with two hidden layers. They did not work better. The [4, 2] network was the worst, because its second layer has only 2 nodes and the network must choose between 3 species. This makes a "bottleneck". The [8, 4] network was better, but still slower than one layer with 8 nodes. For this dataset, one hidden layer with 8–16 nodes was the best choice.

### How the number of epochs affects the results?
More epochs gave better accuracy for every network. With 10 epochs, all networks were too weak (underfitting), and the results changed a lot between runs. With 100–200 epochs, most networks reached 97–100% accuracy, and the results were much more stable. 
After the network reached 100%, more epochs did not help. Some results even dropped a little at 200 epochs, but the drop was very small (only a few wrong predictions), so it is probably just random variation.
