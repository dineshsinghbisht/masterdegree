```python
import tensorflow as tf
import numpy as np
tf.keras.backend.set_floatx('float64') # Removes TensorFlow warnings considering the datatypes

## print Tensorflow version
print("TensorFlow version:", tf.__version__)
```

```python
## Scalar i.e. rank 0 tensor
rank_0 = tf.constant(4,tf.int32, name="myscalar")
print("Value of rank_0:", rank_0)
print("Shape of rank_0:", rank_0.shape)
print("Dtype of rank_0:", rank_0.dtype)
print("Numpy array of rank_0:", rank_0.numpy())
```

```python
## VECTOR i.e. rank / dimension 1 tensor
rank_1 = tf.constant([4, 5, 6], tf.int32, name= "myvector") # tf.bool, tf.float32, tf.float64, tf.int32, tf.int64, tf.string
print("Value of rank_1:", rank_1)
print("Shape of rank_1:", rank_1.shape)
print("Dtype of rank_1:", rank_1.dtype)
print("Numpy array of rank_1:", rank_1.numpy())
```

```python
## MATRIX i.e. rank 2 tensor
rank_2 = tf.constant([[1, 2], [3, 4], [5, 6]], tf.int32)
print("Value of rank_2:", rank_2)
print("Shape of rank_2:", rank_2.shape)
print("Dtype of rank_2:", rank_2.dtype)
print("Numpy array of rank_2:", rank_2.numpy())
```

```python
# Create a changeable tensor
my_var = tf.Variable(4, dtype=tf.int32)
print("Value of my_var:", my_var)
print("Numpy array of my_var:", my_var.numpy())

# Update the value inside the TensorFlow object to 10
my_var.assign(10) 

print("Value of my_var:", my_var.numpy()) # Outputs: 10
print("Result of tf.multiply(my_var, 2):", tf.multiply(my_var, 2)) # Outputs: tf.Tensor(20, shape=(), dtype=int32)
print("Numpy array of tf.multiply(my_var, 2):", tf.multiply(my_var, 2).numpy()) # Outputs: 20
```

```python
tf_variable_matrix = tf.Variable([[1, 2, 3]], dtype=tf.int32)
print(tf_variable_matrix)
print(tf_variable_matrix.numpy()) # Outputs: [[1 2 3]]
print(tf.transpose(tf_variable_matrix)) # Outputs: tf.Tensor([[1], [2], [3]], shape=(3, 1), dtype=int32)
print(tf.transpose(tf_variable_matrix).numpy()) # Outputs: [[1], [2], [3]]
```

```python
import tensorflow as tf

# 1. Calculate for a 2x2 Matrix
matrix_2x2 = tf.constant([[4.0, 7.0], 
                          [2.0, 6.0]])

det_2x2 = tf.linalg.det(matrix_2x2)
print("2x2 Determinant:", det_2x2.numpy()) 
# Math check: (4*6) - (7*2) = 24 - 14 = 10.0


# 2. Calculate for a 3x3 Matrix
matrix_3x3 = tf.constant([[6.0, 1.0, 1.0], 
                          [4.0, -2.0, 5.0], 
                          [2.0, 8.0, 7.0]])

det_3x3 = tf.linalg.det(matrix_3x3)
print("3x3 Determinant:", det_3x3.numpy())

print("transpose of 2x2 matrix:", tf.transpose(matrix_2x2).numpy())
print("transpose of 3x3 matrix:", tf.transpose(matrix_3x3).numpy())
```
