```python
import tensorflow as tf
import numpy as np
tf.keras.backend.set_floatx('float64') # Removes TensorFlow warnings considering the datatypes

## print Tensorflow version
print("TensorFlow version:", tf.__version__)
```

```python
## Scalar i.e. rank 0 tensor
rank_0 = tf.constant(4,tf.int32)
print(rank_0)
print(rank_0.shape)
print(rank_0.numpy())
```

```python
## VECTOR i.e. rank 1 tensor
rank_1 = tf.constant([4, 5, 6], tf.int32) # tf.bool, tf.float32, tf.float64, tf.int32, tf.int64, tf.string
print(rank_1)
print(rank_1.shape)
print(rank_1.numpy())
```

```python
## MATRIX i.e. rank 2 tensor
rank_2 = tf.constant([[1, 2], [3, 4], [5, 6]], tf.int32)
print(rank_2)
print(rank_2.shape)
print(rank_2.numpy())
```

```python
# Create a changeable tensor
my_var = tf.Variable(4, dtype=tf.int32)
print(my_var)
print(my_var.numpy())
# Update the value inside the TensorFlow object to 10
my_var.assign(10) 

print(my_var.numpy()) # Outputs: 10
print(tf.multiply(my_var, 2)) # Outputs: tf.Tensor(20, shape=(), dtype=int32)
print(tf.multiply(my_var, 2).numpy()) # Outputs: 20
```

```python
tf_variable_vector = tf.Variable([1, 2, 3], dtype=tf.int32)
print(tf_variable_vector)
print(tf_variable_vector.numpy()) # Outputs: [1 2 3]
print(tf.transpose(tf_variable_vector)) # Outputs: tf.Tensor([1 2 3], shape=(3,), dtype=int32)
print(tf.transpose(tf_variable_vector).numpy()) # Outputs: [1 2 3]
```

```python
import tensorflow as tf

# 1. Calculate for a 2x2 Matrix
matrix_2x2 = tf.constant([[4.0, 7.0], 
                          [2.0, 6.0]])

det_2x2 = tf.linalg.det(matrix_2x2)
print("2x2 Determinant:", det_2x2.numpy()) 
# Math check: (4*6) - (7*2) = 24 - 14 = 10.0


# 2. Calculate for a 3x3 Matrix
matrix_3x3 = tf.constant([[6.0, 1.0, 1.0], 
                          [4.0, -2.0, 5.0], 
                          [2.0, 8.0, 7.0]])

det_3x3 = tf.linalg.det(matrix_3x3)
print("3x3 Determinant:", det_3x3.numpy())
```
