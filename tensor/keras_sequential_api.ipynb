```python
import pandas as pd 
import numpy as np
import tensorflow as tf
tf.keras.backend.set_floatx('float64') # Removes the TensorFlow warnings considering the Data Types
df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data',header=None,
                 names=['Sepal Length','Sepal Width','Petal Length','Petal Width','Class'])
df
```

```python
df['Class'] = pd.Categorical(df['Class'])  # https://pandas.pydata.org/pandas-docs/stable/user_guide/categorical.html
df['Class'] = df['Class'].cat.codes
df
```
