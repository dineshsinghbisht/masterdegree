```python
import pandas as pd 
import numpy as np
import tensorflow as tf
tf.keras.backend.set_floatx('float64') # Removes the TensorFlow warnings considering the Data Types
df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data',header=None,
                 names=['Sepal Length','Sepal Width','Petal Length','Petal Width','Class'])
df
```

```python
df['Class'] = pd.Categorical(df['Class'])  # https://pandas.pydata.org/pandas-docs/stable/user_guide/categorical.html
df['Class'] = df['Class'].cat.codes
df
```

```python
#tf.keras.set_random_seed(42) # For reproducibility
tf.keras.utils.set_random_seed(42) # For reproducibility
tf.keras.backend.set_floatx('float64') # Removes the TensorFlow warnings considering the Data Types
```

```python
model_sequential = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),
    #tf.keras.InputLayer(input_shape=(4,)),
    # tf.keras.layers.Dense(10, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(10, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])
```

```python
inputs = tf.keras.Input(shape=(4,))
hidden = tf.keras.layers.Dense(10, activation='relu')(inputs)
outputs = tf.keras.layers.Dense(3, activation='softmax')(hidden)
model_functional = tf.keras.Model(inputs=inputs, outputs=outputs)
```

```python
print(model_sequential.summary())
```

```python
print(model_functional.summary())
```

```python
# optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)
# loss_fn = tf.keras.losses.SparseCategoricalCrossentropy()
# accuracy_metric = tf.keras.metrics.SparseCategoricalAccuracy()

model_sequential.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
#model_functional.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model_functional.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001), loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
history_sequential = model_sequential.fit(df.iloc[:,0:4], df.iloc[:,4], epochs=10, verbose=2)
```

```python
history_functional = model_functional.fit(df.iloc[:,0:4], df.iloc[:,4], epochs=10, verbose=2)
```

```python
evaluation_sequential = model_sequential.evaluate(df.iloc[:,0:4], df.iloc[:,4], verbose=2, return_dict=True)
```

```python
evaluation_functional = model_functional.evaluate(df.iloc[:,0:4], df.iloc[:,4], verbose=2)
```

```python
predictions_sequential = model_sequential.predict(df.iloc[:,0:4])
```
